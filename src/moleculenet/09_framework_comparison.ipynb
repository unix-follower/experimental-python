# Framework Comparison for BBBP, Delaney, and Lipophilicity

This notebook consolidates the first baseline results into one place so framework and dataset tradeoffs are visible before adding more model complexity.

Current comparison scope:
- BBBP classification across scikit-learn, PyTorch, and TensorFlow
- Delaney regression across scikit-learn, PyTorch, and TensorFlow
- Lipophilicity regression extension under the same train-valid-test split contract

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## BBBP Classification Metrics

The metrics below use the same random split definition used in the baseline notebooks.

```python
bbbp_results = pd.DataFrame(
    [
        {
            "framework": "scikit-learn",
            "accuracy": 0.8762,
            "f1": 0.9159,
            "roc_auc": 0.9471,
        },
        {"framework": "PyTorch", "accuracy": 0.7296, "f1": 0.8019, "roc_auc": 0.8103},
        {
            "framework": "TensorFlow",
            "accuracy": 0.8241,
            "f1": 0.8916,
            "roc_auc": 0.8546,
        },
    ]
)
display(bbbp_results.round(4))
```

```python
plot_df = bbbp_results.melt(id_vars="framework", var_name="metric", value_name="value")
plt.figure(figsize=(9, 4))
sns.barplot(data=plot_df, x="metric", y="value", hue="framework")
plt.ylim(0.6, 1.0)
plt.title("BBBP Validation Metrics by Framework")
plt.ylabel("Score")
plt.show()
```

## Delaney Regression Metrics

The regression models all use the same descriptor columns and the same train-valid-test split contract.

```python
delaney_results = pd.DataFrame(
    [
        {
            "framework": "scikit-learn ridge",
            "rmse": 0.8650,
            "mae": 0.6414,
            "r2": 0.8515,
        },
        {
            "framework": "scikit-learn random forest",
            "rmse": 0.6137,
            "mae": 0.4316,
            "r2": 0.9252,
        },
        {"framework": "PyTorch MLP", "rmse": 0.6656, "mae": 0.4771, "r2": 0.9120},
        {"framework": "TensorFlow MLP", "rmse": 0.6523, "mae": 0.4588, "r2": 0.9155},
    ]
)
display(delaney_results.round(4))
```

```python
regression_plot_df = delaney_results.melt(
    id_vars="framework", var_name="metric", value_name="value"
)
plt.figure(figsize=(10, 4))
sns.barplot(data=regression_plot_df, x="metric", y="value", hue="framework")
plt.title("Delaney Validation Metrics by Framework")
plt.ylabel("Value")
plt.show()
```

## Regression Expansion: Lipophilicity

Lipophilicity extends the regression track with a SMILES-only text representation rather than explicit descriptor columns. The split contract is the same, but the input modality changes.

```python
lipophilicity_results = pd.DataFrame(
    [
        {
            "dataset": "Lipophilicity",
            "model": "ridge",
            "rmse": 0.8099,
            "mae": 0.6340,
            "r2": 0.5520,
        },
        {
            "dataset": "Lipophilicity",
            "model": "elasticnet",
            "rmse": 0.8493,
            "mae": 0.6650,
            "r2": 0.5074,
        },
    ]
)
display(lipophilicity_results.round(4))
```

```python
dataset_comparison = pd.DataFrame(
    [
        {
            "dataset": "Delaney",
            "best_model": "scikit-learn random forest",
            "rmse": 0.6137,
            "mae": 0.4316,
            "r2": 0.9252,
            "input_style": "tabular descriptors",
        },
        {
            "dataset": "Lipophilicity",
            "best_model": "scikit-learn ridge",
            "rmse": 0.8099,
            "mae": 0.6340,
            "r2": 0.5520,
            "input_style": "SMILES text n-grams",
        },
    ]
)
display(dataset_comparison.round(4))
```

```python
dataset_plot_df = dataset_comparison.melt(
    id_vars=["dataset", "best_model", "input_style"],
    var_name="metric",
    value_name="value",
)
plt.figure(figsize=(9, 4))
sns.barplot(data=dataset_plot_df, x="metric", y="value", hue="dataset")
plt.title("Best Regression Baselines by Dataset")
plt.ylabel("Value")
plt.show()
```
